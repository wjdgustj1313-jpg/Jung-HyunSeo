```python
import matplotlib.pyplot as plt
import numpy as np

joint_pmf_data = {
    (1, 1): 0.1,
    (1, 2): 0.2,
    (1, 3): 0.2,
    (2, 1): 0.3,
    (2, 2): 0.05,
    (2, 3): 0.15
}

f_X = {}
f_Y = {}

for (x, y), p in joint_pmf_data.items():
    f_X[x] = f_X.get(x, 0) + p

for (x, y), p in joint_pmf_data.items():
    f_Y[y] = f_Y.get(y, 0) + p

plt.figure(figsize=(12, 5))

y_min = 0
y_max = 1.0
x_min = 0.5
x_max = 3.5

plt.subplot(1, 2, 1)

x_values_X = sorted(f_X.keys())
y_values_X = [f_X[x] for x in x_values_X]

plt.plot(x_values_X, y_values_X, 'bo', label='P(X=x)')
plt.vlines(x_values_X, y_min, y_values_X, colors='blue', lw=3)

plt.title('Marginal PMF of X', fontsize=14)
plt.xlabel('X value', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.xticks(np.arange(int(x_min) + 0.5, int(x_max)), fontsize=10)
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(axis='y', linestyle='--')
plt.legend()

plt.subplot(1, 2, 2)

y_values_Y = sorted(f_Y.keys())
y_values_Y_prob = [f_Y[y] for y in y_values_Y]

plt.plot(y_values_Y, y_values_Y_prob, 'ro', label='P(Y=y)')
plt.vlines(y_values_Y, y_min, y_values_Y_prob, colors='red', lw=3)

plt.title('Marginal PMF of Y', fontsize=14)
plt.xlabel('Y value', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.xticks(np.arange(int(x_min) + 0.5, int(x_max)), fontsize=10)
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(axis='y', linestyle='--')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np

korean_scores = [75, 65, 50, 70, 80]
math_scores = [65, 70, 70, 55, 60]

data = np.array([korean_scores, math_scores])

covariance = np.cov(data, bias=True)[0, 1]

print("Covariance:", covariance)
```
